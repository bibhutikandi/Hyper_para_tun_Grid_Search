```python
from scikeras.wrappers import KerasClassifier
import pandas as pd
import numpy as np
from sklearn.model_selection import GridSearchCV
import warnings
warnings.filterwarnings('ignore')
from tensorflow.keras.datasets import mnist
import tensorflow as tf
```

```python
(x_train,y_train),(x_test,y_test)= mnist.load_data()
x_train.shape
x_test.shape
y_train  .shape
y_test.shape
```

```python
x_train= x_train.reshape(60000,784)
x_test= x_test.reshape(10000,784)
x_train=x_train/255
x_test=x_test/255
```

```python
y_train= tf.keras.utils.to_categorical(y_train)
y_test= tf.keras.utils.to_categorical(y_test)
```

```python
def create_my_model1():
    model=tf.keras.models.Sequential()
    model.add(tf.keras.layers.Dense(560,activation='relu',input_dim=(784)))
    model.add(tf.keras.layers.Dense(450,activation='relu'))
    model.add(tf.keras.layers.Dense(10,activation='softmax'))
    model.compile(optimizer=tf.keras.optimizers.Adam(),loss='categorical_crossentropy',metrics=['accuracy'])
    return model
```

```python
model1=KerasClassifier(build_fn=create_my_model1)
```

```python
batch_values=[100,200]
epochs_value=[10,5]
```

```python
params= dict(batch_size=batch_values,epochs=epochs_value)
params
```

```python
grid_search= GridSearchCV(estimator=model1,param_grid=params,cv=2)
```

```python
grid_search.fit(x_train,y_train)
```

```python
## summary_results
grid_search.best_score_, grid_search.best_params_
```

```python
optimizer_types=['SGD','Adagrad','RMSprop','Adam']
param_grid= dict(optimizer=optimizer_types)
param_grid
```

```python
def create_my_model(optimizer='SGD'):
    model=tf.keras.models.Sequential()
    model.add(tf.keras.layers.Dense(560,activation='relu',input_dim=(784)))
    model.add(tf.keras.layers.Dense(450,activation='relu'))
    model.add(tf.keras.layers.Dense(10,activation='softmax'))
    model.compile(optimizer=optimizer,loss='categorical_crossentropy',metrics=['accuracy'])
    return model
```

```python
model2=KerasClassifier(build_fn= create_my_model,epochs=10,batch_size=100)
```

```python
grid_search2= GridSearchCV(estimator=model2, param_grid= param_grid,cv=3)
grid_search2.fit(x_train,y_train)
```

```python
grid_search2.best_score_,grid_search2.best_params_
```
